```python
# Install required libraries (run this once)
# !pip install pandas numpy scikit-learn scikit-surprise

import pandas as pd
from surprise import SVD, Dataset, Reader, accuracy
from surprise.model_selection import train_test_split


# Load Data

def load_data():
    df = pd.read_csv('/content/Last.fm_data.csv')

    # Optional: limit for faster runs
    df = df.head(50000)

    print("Sample data:")
    print(df.head())

    # Aggregate play counts (Username × Track)
    play_counts = df.groupby(['Username', 'Track']).size().reset_index(name='Playcount')
    return play_counts

play_counts = load_data()


# Train SVD Model

def train_model(data):
    # Define rating scale based on observed playcount range
    reader = Reader(rating_scale=(0, data['Playcount'].max()))

    # Convert to Surprise dataset format
    dataset = Dataset.load_from_df(data[['Username', 'Track', 'Playcount']], reader)

    # Split dataset into training and testing sets
    trainset, testset = train_test_split(dataset, test_size=0.2)

    # Train the SVD model
    model = SVD()
    model.fit(trainset)

    # Evaluate model performance
    predictions = model.test(testset)
    rmse = accuracy.rmse(predictions, verbose=True)
    print(f"\n✅ Model trained successfully. RMSE = {rmse:.4f}\n")

    return model

model = train_model(play_counts)

# Recommendation Function

def recommend_songs(user_id, n=5):
    all_songs = play_counts['Track'].unique()
    listened_songs = play_counts[play_counts['Username'] == user_id]['Track'].values

    # Predict ratings for songs the user hasn't listened to
    predictions = []
    for song in all_songs:
        if song not in listened_songs:
            est = model.predict(user_id, song).est
            predictions.append((song, est))

    # Sort by predicted rating (descending)
    predictions.sort(key=lambda x: x[1], reverse=True)

    return predictions[:n]

# Example: Get Recommendations

sample_user = play_counts['Username'].iloc[0]
top_n = 5

recommendations = recommend_songs(sample_user, n=top_n)

print(f"🎵 Top {top_n} Recommended Songs for User: {sample_user}")
for i, (track, score) in enumerate(recommendations, start=1):
    print(f"{i}. {track} (Predicted Rating: {score:.2f})")
```
